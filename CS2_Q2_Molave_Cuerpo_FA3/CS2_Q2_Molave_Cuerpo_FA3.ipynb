```python
scores = [85, 92, 78, 90, 88]


def display_scores():
    print("\nStudent scores:")
    for index, score in enumerate(scores):
        print(f"Student {index + 1}: {score}")


def search_score():
    try:
        target = int(input("Enter the score to search for: "))
    except ValueError:
        print("Please enter a valid whole number.")
        return

    matching_students = [
        index + 1 for index, score in enumerate(scores) if score == target
    ]

    if matching_students:
        print(f"Score {target} found for student(s): {matching_students}")
    else:
        print(f"Score {target} was not found.")


def compute_average():
    if not scores:
        print("There are no scores to average.")
        return

    average = sum(scores) / len(scores)
    print(f"Average score: {average:.2f}")


def update_score():
    display_scores()

    try:
        student_number = int(input("Enter the student number to update: "))
        new_score = int(input("Enter the new score: "))
    except ValueError:
        print("Please enter valid whole numbers.")
        return

    index = student_number - 1

    if 0 <= index < len(scores):
        old_score = scores[index]
        scores[index] = new_score  # Update the value using its index
        print(f"Student {student_number}'s score changed from {old_score} to {new_score}.")
    else:
        print("Invalid student number.")


def main():
    while True:
        print("\n=== Student Score Analyzer ===")
        print("1. Display Scores")
        print("2. Search Score")
        print("3. Compute Average Score")
        print("4. Update Score")
        print("5. Display Updated Scores")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            display_scores()
        elif choice == "2":
            search_score()
        elif choice == "3":
            compute_average()
        elif choice == "4":
            update_score()
        elif choice == "5":
            display_scores()
        elif choice == "6":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Please select a number from 1 to 6.")


main()
```
